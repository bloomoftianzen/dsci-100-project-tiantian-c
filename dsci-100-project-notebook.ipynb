Visualization 1: A scatter plot investigating the relationship between conference and regional win percentages is relevant to our analysis. Because conference games precede regional competitions, examining performance across schools over time allows us to explore whether pre-game performance predicts performance in regional competitions. This will also provide us with insights on whether pre-game performance could be the indicators for actual performance during the tournaments.

```r
### 
conf_reg_vis <- ggplot(women_basketball_tidy_data, aes(x = conf_wins_pct,
                                                       y = reg_wins_pct)) +
                geom_point()+
                labs(title = "Conference Win Percentage vs. Regional Win Percentage over 1982-2018",
                x = "Conference Win Percentage",
                y = "Regional Win Percentage") 

conf_reg_vis
###
```

Visualization 2: This line plot of average regional win percentages over time (1982–2018) is relevant because it lets us check if teams are generally getting stronger or weaker across the decades. Since the graph shows that average win percentages are fluctuating over time without major trends, it helps us confirm that time isn't confounding with our data. This allows us to safely focus our exploratory analysis on how pre-tournament performances, such as: conference and regional win percentages relate to the teams' final tournament standings.

```r
### 
year_reg_trend_vis <- women_basketball_tidy_data |>
                    group_by(year) |>
                summarize(mean_reg_wins = mean(reg_wins_pct, na.rm = TRUE)) |>
                ggplot(aes(x = year,y = mean_reg_wins)) +
                geom_line()+
                labs(title = "Average Regional Win Percentage over Time(1982-2018)",
                x = "Year(1982-2018)",
                y = "Average Regional Win Percentage across all schools in a given year") 

year_reg_trend_vis
###
```
